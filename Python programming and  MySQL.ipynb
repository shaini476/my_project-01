```python
import mysql.connector
```

```python
# Establishing the connection
mydb = mysql.connector.connect(              
                         host = 'localhost',
                         user = 'root',
                         password = 'shaini@18111')
```

```python
# Creating a cursor object using the cursor() method
cursor = mydb.cursor()
```

```python
# # Create the database query
# query="""CREATE DATABASE demo10;"""

# # Executing the creating database query
# cursor.execute(query)
```

```python
# Display the list of databases
cursor.execute('SHOW DATABASES')
for database in cursor:
    print(database[0])
```

```python
# USE() method is use to take the existing database 'demo10'
database_query = "USE demo10"
cursor.execute(database_query)
```

```python
# using the DROP statement to delete the DEPARTMENT table if already exist
delete_table_query="""DROP TABLE department;"""

# Executing the delete table query
cursor.execute(delete_table_query)
```

```python
# Creating table DEPARTMENT using constraints whereever necessary
query = """ CREATE TABLE department(
                    deptid int NOT NULL UNIQUE,
                    deptname VARCHAR(50) NOT NULL)"""
                    
# Executing the creating table query
cursor.execute(query)
```

```python
# Display the list of table
cursor.execute('SHOW TABLES')
for table in cursor:
    print(table[0])
```

```python
# preparing query1 to insert a record into table DEPARTMENT
query1 = """INSERT INTO department (deptid, deptname)
            VALUES (101,'Human Resources' )"""

# use the cursor to execute a query1
cursor.execute(query1)
```

```python
# preparing query2 to insert a record into table DEPARTMENT
query2 = """ INSERT INTO department (deptid, deptname)
            VALUES (%s, %s) """              # %s as placeholder

val = [
    (102, 'Information Technology'),
    (103, 'Finance'),
    (104, 'Sales'),
    (105, 'Administration'),
    (106, 'Executive'),
    (107, 'Marketing')
]

# To insert multiple rows into a table DEPARTMENT ,using the executemany() method
cursor.executemany(query2,val)
```

```python
# SELECT statement is used to get the result what we want from the table DEPARTMENT
query3 = "SELECT * FROM department"

# Executing the command query3
cursor.execute(query3)
# using fetchall() method to fetch all row from the table DEPARTMENT
result = cursor.fetchall()

# display the department information
print("Department Info : ")
# iterating through the row
for row in result:
    print(row)
```

```python
# UPDATE command is used to update the value of table DEPARTMENT
query4 = """UPDATE department
            SET deptname = 'Marketing'
            WHERE deptid = 104"""

# Executing the command query4
cursor.execute(query4)
cursor.execute('SELECT * FROM department')
# using fetchall() method to fetch all row from the table DEPARTMENT
result = cursor.fetchall()

# display the department information
print("Department info : ")
# iterating through the row
for row in result:
    print(row)
```

```python
# DELETE command is used to delete the row of table DEPARTMENT
query5 = """ DELETE FROM department
             WHERE deptname = 'Marketing' """

# Executing the command query5
cursor.execute(query5)
cursor.execute('SELECT * FROM department')
# using fetchall() method to fetch all row from the table DEPARTMENT
result = cursor.fetchall()

# display the department information
print("Department info : ")
# iterating through the row
for row in result:
    print(row)
```

```python
# commit changes in the databases
mydb.commit()
# close() method is used to close the connection to the database
mydb.close()
```
